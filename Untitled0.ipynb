## تعريف الدوال

```python
with open("weblogs.txt", "r") as f:
    lines = f.readlines()

def mapper(lines):
    mapped = []
    for line in lines:
        if line.startswith("#"):
            continue
        parts = line.strip().split(",")
        status = parts[5]
        mapped.append((status, 1))
    return mapped

def group(mapped):
    grouped = {}
    for key, value in mapped:
        if key not in grouped:
            grouped[key] = []
        grouped[key].append(value)
    return grouped

def reducer(grouped):
    result = {}
    for key in grouped:
        result[key] = sum(grouped[key])
    return result

mapped = mapper(lines)
grouped = group(mapped)
result = reducer(grouped)

for key, value in result.items():
    print(f"HTTP {key}: {value} requests")
```

## عدد الطلبات لكل URL

```python
def mapper_url(lines):
    mapped = []

    for line in lines:
        if line.startswith("#"):
            continue

        parts = line.strip().split(",")
        url = parts[4]   # استخراج URL

        mapped.append((url, 1))

    return mapped


mapped = mapper_url(lines)
grouped = group(mapped)
result = reducer(grouped)

print("Requests per URL:")
for key, value in result.items():
    print(f"{key}: {value}")
```

## مجموع حجم البيانات لكل Status

```python
def mapper_size(lines):
    mapped = []

    for line in lines:
        if line.startswith("#"):
            continue

        parts = line.strip().split(",")
        status = parts[5]
        size = int(parts[6])   # حجم البيانات

        mapped.append((status, size))

    return mapped


mapped = mapper_size(lines)
grouped = group(mapped)
result = reducer(grouped)

print("\nTotal Response Size per Status:")
for key, value in result.items():
    print(f"HTTP {key}: {value} bytes")
```

## تحليل الأخطاء فقط

```python
def mapper_errors(lines):
    mapped = []

    for line in lines:
        if line.startswith("#"):
            continue

        parts = line.strip().split(",")
        status = parts[5]

        if status != "200":   # فلترة الأخطاء
            mapped.append((status, 1))

    return mapped


mapped = mapper_errors(lines)
grouped = group(mapped)
result = reducer(grouped)

print("\nErrors only:")
for key, value in result.items():
    print(f"HTTP {key}: {value} requests")
```
